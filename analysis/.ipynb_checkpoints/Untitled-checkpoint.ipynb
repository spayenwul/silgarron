```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Добавляем корневую папку проекта в путь, чтобы можно было импортировать наши модули
# ИЗМЕНЕНИЕ ЗДЕСЬ: .parent говорит "взять родительскую папку"
project_root = Path.cwd().parent 
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))
    print(f"Добавили '{project_root}' в sys.path")

# Импортируем наши сервисы
from services.world_data_service import WorldDataService
from services.compatibility_service import CompatibilityService

# Эта "магическая" команда говорит Jupyter отображать графики прямо в ноутбуке
%matplotlib inline

print("✅ Окружение готово, сервисы импортированы.")
```

```python
# Создаем экземпляр сервиса, ПЕРЕДАВАЯ ему путь к корню проекта
wds = WorldDataService(project_root)

# Остальная логика не меняется, так как она зависит от правильно созданного wds
rules = wds.get_generation_rules()
cs = CompatibilityService(rules)

print("✅ Сервисы инициализированы, данные мира загружены.")
```

```python
# --- Ячейка 3: Вычисляем матрицу совместимости ---

def generate_compatibility_matrix(comp_service: CompatibilityService, world_data: WorldDataService) -> dict:
    """Генерирует матрицу совместимости на основе загруженных данных."""
    matrix = {}
    
    all_races = world_data.get_all_races()
    all_biomes = world_data.get_all_biomes()

    if not all_races:
        print("⚠️ Предупреждение: Расы не найдены.")
        return {}

    race_ids = [r['id'] for r in all_races]
    
    for race_id in race_ids:
        matrix[race_id] = {}
        for biome_data in all_biomes:
            biome_id = biome_data['id']
            # Находим полные данные расы по ее ID
            race_data = next((r for r in all_races if r['id'] == race_id), None)
            if race_data:
                score = comp_service.calculate_race_biome_score(race_data, biome_data)
                matrix[race_id][biome_id] = score.raw_score
            
    return matrix

# Вызываем функцию с нашими сервисами, созданными в предыдущей ячейке
comp_matrix = generate_compatibility_matrix(cs, wds)

# ВАЖНО: Просто пишем имя переменной в конце.
# Jupyter автоматически красиво отобразит ее содержимое под ячейкой.
comp_matrix
```

```python
df = pd.DataFrame(comp_matrix).T
race_order = df.mean(axis=1).sort_values().index
biome_order = df.mean(axis=0).sort_values().index
df_sorted = df.loc[race_order, biome_order]

plt.figure(figsize=(20, 12))
sns.heatmap(df_sorted, annot=True, fmt='.2f', cmap='RdYlGn', center=1.0)
plt.title("Race-Biome Compatibility (Sorted)", fontsize=16)
plt.xlabel('Biomes')
plt.ylabel('Races'); # точка с запятой подавляет вывод текста вроде <matplotlib.axes._subplots.AxesSubplot at 0x...>
```

### Анализ Тепловой Карты Совместимости

1.  **Общий вывод:** Карта показывает хороший разброс значений, от низких (красные) до высоких (зеленые). Это значит, что правила совместимости работают и создают контраст.
2.  **Самые адаптивные расы:** Судя по сортировке, расы `[имя расы]` и `[имя расы]` (последние строки) являются самыми универсальными.
3.  **Самые гостеприимные биомы:** `[имя биома]` и `[имя биома]` (правые столбцы) в среднем лучше подходят для жизни.
4.  **Аномалии:** Виден интересный паттерн, где раса `[имя расы]` идеально подходит для `[имя биома]`, но проваливается везде еще. Это отличный кандидат на расу-"специалиста".

```python
# Собираем все значения из матрицы в один список
all_scores = df.values.flatten()

plt.figure(figsize=(10, 6))
sns.histplot(all_scores, bins=20, kde=True)
plt.title('Distribution of All Compatibility Scores')
plt.xlabel('Compatibility Score')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.5)
```

### Анализ Распределения

Гистограмма показывает два явных пика:
- **Пик в районе 0.2-0.4:** Это наши правила "несовместимости". Хорошо, что они работают.
- **Основной пик в районе 1.0-1.5:** Это нейтральные и умеренно-хорошие сочетания.
- **Длинный "хвост" до 2.5:** Это наши правила синергии. Они создают уникальные, "идеальные" сочетания, но не слишком часто.

**Вывод:** Баланс системы выглядит адекватным. Нет перекоса в одну сторону.

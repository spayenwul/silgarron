### 2.2 Распределение очков совместимости

**Вопрос**: Создаёт ли моя система осмысленный контраст между биомами?

**Интерпретация**:
- **Высокая вариативность (std > 0.5)**: Система создаёт осмысленные различия
- **Низкая вариативность (std < 0.3)**: Правила слишком слабые, все биомы одинаково совместимы
- **Несколько пиков**: Хорошо — есть группы очень совместимых и очень несовместимых биомов

```python
# Сначала мы должны создать сервисы, которые анализатор будет использовать
from services.world_data_service import WorldDataService
from services.compatibility_service import CompatibilityService
import yaml

# 1. Создаем WorldDataService, передавая ему корень проекта
wds = WorldDataService(project_root_path=PROJECT_ROOT)

# 2. Загружаем правила, как это делает анализатор, чтобы создать CompatibilityService
tags_registry_path = PROJECT_ROOT / "data" / "tags_registry.yaml"
with open(tags_registry_path, 'r', encoding='utf-8') as f:
    tags_registry = yaml.safe_load(f)

generation_rules_path = PROJECT_ROOT / "data" / "generation_rules.yaml"
with open(generation_rules_path, 'r', encoding='utf-8') as f:
    generation_rules = yaml.safe_load(f)

cs = CompatibilityService(
    tags_registry=tags_registry,
    generation_rules=generation_rules
)

# 3. Теперь создаем анализатор, ПЕРЕДАВАЯ ему уже готовые сервисы
#    (Вам нужно будет сначала изменить __init__ в compatibility_analyzer.py, как мы обсуждали)
compat_analyzer = CompatibilityAnalyzer(wds=wds, cs=cs)

# 4. Теперь этот вызов должен работать без ошибок
compat_analyzer.plot_compatibility_distribution(
    save_path = results_dir / "compatibility_distribution.png" # Рекомендую сохранять в папку results
)
```

### 2.3 Специализация рас

**Вопрос**: Есть ли у меня расы-специалисты и расы-универсалы?

**Интерпретация Box Plot**:
- **Короткий box (узкий диапазон)**: Раса-универсал, одинаково приспособлена везде

### 3.2 Матрица смежности биомов

**Вопрос**: Какие биомы часто граничат друг с другом?

**Интерпретация**:
- Высокие значения → частое соседство (возможно, синергия)
- Низкие значения → редкое соседство (возможно, конфликт или запрет)
- Асимметрия матрицы может указывать на направленные правила

```python
#Определяем путь для сохранения
save_path = results_dir / "adjacency_matrix_demo.png"
#Передаем путь в метод
gen_analyzer.analyze_adjacency_matrix(world_graph=None, save_path=save_path)
```

**Вопрос**: Как выглядит топология мира?

**Что показывает граф**:
- **Изолированные зоны**: Биомы без соседей (возможная проблема)
- **Центральные хабы**: Биомы с множеством соседей (универсальные)
- **Кластеры**: Группы биомов, которые часто встречаются вместе

## 6. Интерпретация и рекомендации

### Что делать с результатами:

#### Если распределение совместимости показывает низкую вариативность (std < 0.3):
1. Увеличьте значения `bonus` в синергиях
2. Увеличьте `penalty` (уменьшите множитель) в конфликтах
3. Добавьте больше тегов для контрастных биомов

#### Если раса слишком универсальна (нужна специализация):
1. Добавьте расе больше специфичных тегов
2. Создайте синергии для её нишевых тегов
3. Добавьте конфликты с противоположными биомами

#### Если раса слишком специализирована (нужна адаптивность):
1. Добавьте расе универсальные теги
2. Уменьшите штрафы в конфликтах
3. Добавьте бонусы к `race_modifiers` в конфиге расы

#### Если биомы редко появляются:
1. Проверьте `spawn_weight` биома
2. Проверьте, не слишком ли строгие правила совместимости
3. Убедитесь, что биом не в `forbidden_combinations`

### Следующие шаги:
1. После реализации `SpatialLocationGenerator` замените `generator=None`
2. Запустите real-world генерацию и сравните с DEMO данными
